```python
import numpy as np
import matplotlib.pyplot as plt

# Given values 
peak_height_mm = 8848 * 1000       # Everest height, converted to mm
paper_thick_mm = 0.1
tissue_thick_mm = 0.05
card_thick_mm = 0.3
```

```python
 #(a) Folds needed to exceed Everest's height with a standard paper

folds_for_paper = np.ceil(np.log2(peak_height_mm / paper_thick_mm))
print("Standard paper folds:", int(folds_for_paper))
```

```python
# (b) Comparing folds required across the different paper types
folds_for_tissue = np.ceil(np.log2(peak_height_mm / tissue_thick_mm))
folds_for_card = np.ceil(np.log2(peak_height_mm / card_thick_mm))
print("Tissue paper folds:", int(folds_for_tissue))
print("Cardstock folds:", int(folds_for_card))
```

```python
# (c) Thickness that would require exactly 20 folds
# Inverting the formula
target_fold_count = 20
thick_for_20_folds = peak_height_mm / (2 ** target_fold_count)
print("Thickness needing exactly 20 folds:", round(thick_for_20_folds, 4), "mm")
```

```python
# (d) plot ofFolds required as a function of paper thickness
thick_list_mm = np.logspace(np.log10(0.02), np.log10(1.0), 100)
fold_list = np.ceil(np.log2(peak_height_mm / thick_list_mm))

plt.plot(thick_list_mm, fold_list, label="Folds needed")
plt.scatter(paper_thick_mm, folds_for_paper, color="red")
plt.annotate("standard paper", (paper_thick_mm, folds_for_paper))
plt.scatter(tissue_thick_mm, folds_for_tissue, color="green")
plt.annotate("tissue paper", (tissue_thick_mm, folds_for_tissue))
plt.scatter(card_thick_mm, folds_for_card, color="orange")
plt.annotate("cardstock", (card_thick_mm, folds_for_card))
plt.xscale("log")
plt.xlabel("Paper thickness (mm, log scale)")
plt.ylabel("Number of folds needed")
plt.title("Folds Needed to Exceed Mount Everest's Height vs Paper Thickness")
plt.legend()
plt.grid(True, which="both")
plt.show()
```

```python
# Steps 1 & 2: Setting the initial volume and time counter
import numpy as np
import matplotlib.pyplot as plt

start_volume = 1000.0     
step_size = 0.001         
```

```python
# Step 3: finding the time to reach 50% and 10% (a = 0.05)
def search_time_for_threshold(v0, decay_rate, target_fraction, dt):
    volume_now = v0
    time_now = 0.0
    # step forward in time until volume fraction drops below the target
    while volume_now / v0 >= target_fraction:
        time_now += dt
        volume_now = v0 * np.exp(-decay_rate * time_now)
    return time_now

decay_rate_a = 0.05       
empty_fraction = 0.01

time_half_a = search_time_for_threshold(start_volume, decay_rate_a, 0.5, step_size)
time_low_a = search_time_for_threshold(start_volume, decay_rate_a, 0.1, step_size)
time_empty_a = search_time_for_threshold(start_volume, decay_rate_a, empty_fraction, step_size)

print("a = 0.05")
print("Time to reach 50%:", round(time_half_a, 2), "days")
print("Time to reach 10%:", round(time_low_a, 2), "days")
print("Time to reach 1%:", round(time_empty_a, 2), "days")
```

```python
# Step 5: Repeat steps 3 and 4 at a = 0.15
decay_rate_b = 0.15

time_half_b = search_time_for_threshold(start_volume, decay_rate_b, 0.5, step_size)
time_low_b = search_time_for_threshold(start_volume, decay_rate_b, 0.1, step_size)
time_empty_b = search_time_for_threshold(start_volume, decay_rate_b, empty_fraction, step_size)

print("a = 0.15")
print("Time to reach 50%:", round(time_half_b, 2), "days")
print("Time to reach 10%:", round(time_low_b, 2), "days")
print("Time to reach empty (1%):", round(time_empty_b, 2), "days")
```

```python
# Step 6: Plot both decay curves with all 6 threshold points marked
day_range = np.linspace(0, 100, 500)
fraction_curve_a = np.exp(-decay_rate_a * day_range)
fraction_curve_b = np.exp(-decay_rate_b * day_range)

plt.plot(day_range, fraction_curve_a, label="a = 0.05")
plt.plot(day_range, fraction_curve_b, label="a = 0.15")

# mark the three points on curve a
plt.scatter(time_half_a, 0.5, color="black")
plt.scatter(time_low_a, 0.1, color="black")
plt.scatter(time_empty_a, empty_fraction, color="black")

# mark the three points on curve b
plt.scatter(time_half_b, 0.5, color="red")
plt.scatter(time_low_b, 0.1, color="red")
plt.scatter(time_empty_b, empty_fraction, color="red")

plt.xlabel("Time (days)")
plt.ylabel("Volume fraction (v(t) / v0)")
plt.title("Reservoir Volume Decay for Two Different Decay Rates")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 1: Compute compound interest over 5 years 
import numpy as np

principal = 500          # USD
baseline_rate = 0.05    
n = 1                    
years = np.arange(1, 6)  

baseline_balances = principal * (1 + baseline_rate/n) ** (n * years)

for year, balance in zip(years, baseline_balances):
    print(f"Year {year}: ${balance:.2f}")
```

```python
# Step 2: Recompute using real savings rates from two East African banks

bk_rate = 0.05      # Bank of Kigali, student savings
ncba_rate = 0.07    # NCBA, Impano savings account

bk_balances = principal * (1 + bk_rate/n) ** (n * years)
ncba_balances = principal * (1 + ncba_rate/n) ** (n * years)

print("Bank of Kigali balances:", np.round(bk_balances, 2))
print("NCBA balances:", np.round(ncba_balances, 2))
```

```python
# Step 3: Deposit at which the baseline 5% and the best local bank rate differ by more than $100
growth_factor_baseline = (1 + baseline_rate/n) ** (n * 5)
growth_factor_best_local = (1 + ncba_rate/n) ** (n * 5)

minimum_deposit = 100 / abs(growth_factor_best_local - growth_factor_baseline)
print("Minimum deposit for >$100 difference:", round(minimum_deposit, 2))
```

# Question 4

```python
# Step 1: Monthly payment to clear the original car loan
import numpy as np

def loan_monthly_payment(loan_amount, monthly_rate, num_months):
    growth = (1 + monthly_rate) ** num_months
    return loan_amount * monthly_rate * growth / (growth - 1)

car_price = 20000
monthly_rate = 0.05
loan_terms_months = [12, 24, 36]

print("Original car — monthly payments:")
for term in loan_terms_months:
    payment = loan_monthly_payment(car_price, monthly_rate, term)
    print(f"{term} months: ${round(payment)}")
```

```python
# Step 2: Recalculate for the electric car (15% import tax removed)
ev_car_price = car_price * (1 - 0.15)
print("Electric car price:", ev_car_price)

print("\nElectric car — monthly payments:")
for term in loan_terms_months:
    payment = loan_monthly_payment(ev_car_price, monthly_rate, term)
    print(f"{term} months: ${round(payment)}")
```

```python
# Step 3: Savings from switching to the EV, per loan term
print("Total repayment comparison:")
for term in loan_terms_months:
    original_payment = round(loan_monthly_payment(car_price, monthly_rate, term))
    ev_payment = round(loan_monthly_payment(ev_car_price, monthly_rate, term))

    original_total = original_payment * term
    ev_total = ev_payment * term
    total_savings = original_total - ev_total
    monthly_savings = original_payment - ev_payment

    print(f"{term} months -> original total: ${original_total}, EV total: ${ev_total}, "
          f"total savings: ${total_savings}, monthly payment savings: ${monthly_savings}/month")
```

# Question 5

```python
# Step 1 & 2: No-deal scenario — days to repay the $100,000 investment
import numpy as np
import matplotlib.pyplot as plt

def cumulative_profit(profit_per_customer, day_one_customers, growth_rate, day):
    return profit_per_customer * day_one_customers * ((1 + growth_rate) ** day - 1) / growth_rate

investment_original = 100000
profit_per_customer = 5
day_one_customers = 40
growth_no_deal = 0.005

day_counter = 0
profit_so_far = 0
while profit_so_far < investment_original:
    day_counter += 1
    profit_so_far = cumulative_profit(profit_per_customer, day_one_customers, growth_no_deal, day_counter)

breakeven_day_no_deal = day_counter
print("Breakeven day (no deal):", breakeven_day_no_deal)
print("Cumulative profit at that day:", round(profit_so_far, 2))
```

```python
# Step 3: Plotting the  cumulative profit and marking the investment line and breakeven day
days_range = np.arange(1, 400)
profit_curve_no_deal = cumulative_profit(profit_per_customer, day_one_customers, growth_no_deal, days_range)

plt.plot(days_range, profit_curve_no_deal, label="Cumulative profit")
plt.axhline(investment_original, color="gray", linestyle="--", label="Initial investment ($100,000)")
plt.scatter(breakeven_day_no_deal, profit_so_far, color="red", zorder=5, label=f"Breakeven day ({breakeven_day_no_deal})")

plt.xlabel("Day")
plt.ylabel("Cumulative profit ($)")
plt.title("Cumulative Profit vs Day — No Deal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 4: Comparing profit under the deal vs. no deal
growth_with_deal = 0.007
entrepreneur_share = 0.5

days_range = np.arange(1, 600)
profit_no_deal = cumulative_profit(profit_per_customer, day_one_customers, growth_no_deal, days_range)
profit_with_deal = entrepreneur_share * cumulative_profit(profit_per_customer, day_one_customers, growth_with_deal, days_range)

# find the first day the deal's profit overtakes the no-deal profit
crossover_days = days_range[profit_with_deal > profit_no_deal]
crossover_day = crossover_days[0] if len(crossover_days) > 0 else None
print("Day the deal's profit overtakes no-deal profit:", crossover_day)
```

```python
# Step 5: Plotting both profit streams
plt.plot(days_range, profit_no_deal, label="No deal (100% share, 0.5% growth)")
plt.plot(days_range, profit_with_deal, label="With deal (50% share, 0.7% growth)")
plt.scatter(crossover_day, profit_no_deal[crossover_day-1], color="red", zorder=5, label=f"Crossover (day {crossover_day})")

plt.xlabel("Day")
plt.ylabel("Cumulative profit ($)")
plt.title("Cumulative Profit: No Deal vs. Deal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Part (a): Loading data, fixing encoding, aggregate to national totals
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('drc_zones_timeseries-zonal.csv')
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')

def fix_encoding(text):
    # reverses a double UTF-8 encoding artifact 
    try:
        return text.encode('latin1').decode('utf-8')
    except (UnicodeDecodeError, UnicodeEncodeError):
        return text

df['Province'] = df['Province'].apply(fix_encoding)
df['Province'] = df['Province'].replace({'Haut-Uele': 'Haut-Uélé'})  # merge duplicate spelling

print("Provinces affected:", sorted(df['Province'].unique()))
print("Number of provinces:", df['Province'].nunique())
print("Number of health zones:", df['Health Zone'].nunique())
print("Date range:", df['Date'].min().date(), "to", df['Date'].max().date())
```

```python
# National totals and verification against WHO/ECDC
latest_per_zone = df.sort_values('Date').groupby('Health Zone').last()
total_cases = latest_per_zone['Confirmed Cases'].sum()
total_deaths = latest_per_zone['Confirmed Deaths'].sum()

print("Total confirmed cases (as of 2026-08-04):", int(total_cases))
print("Total confirmed deaths:", int(total_deaths))
print("Case fatality rate:", round(100 * total_deaths / total_cases, 1), "%")
```

```python
# Part (b): Checking for missing reporting dates
full_date_range = pd.date_range(df['Date'].min(), df['Date'].max(), freq='D')
reported_dates = set(df['Date'].unique())
missing_dates = [d for d in full_date_range if d not in reported_dates]

print("Total calendar days in range:", len(full_date_range))
print("Days with at least one report:", len(full_date_range) - len(missing_dates))
print("Days with no report at all:", len(missing_dates))

missing_weekday_counts = pd.Series(missing_dates).dt.day_name().value_counts()
print("\nMissing dates by weekday:")
print(missing_weekday_counts)
```

```python
# Part (c): Interpolation methods
def build_clean_national_series(df, value_column):
    pivot = df.pivot_table(index='Date', columns='Health Zone', values=value_column, aggfunc='last')
    pivot = pivot.reindex(full_date_range)

    pivot = pivot.cummax()  # fix any downward data revisions per zone

    for zone in pivot.columns:
        first_report_date = pivot[zone].first_valid_index()
        if first_report_date is not None:
            pivot.loc[pivot.index < first_report_date, zone] = 0  

    pivot = pivot.interpolate(method='linear', limit_direction='forward')
    pivot = pivot.ffill()  # hold last known value after each zone's final report
    return pivot.sum(axis=1)

national_cases = build_clean_national_series(df, 'Confirmed Cases')
national_deaths = build_clean_national_series(df, 'Confirmed Deaths')

print("Cases series non-decreasing:", national_cases.is_monotonic_increasing)
print("Deaths series non-decreasing:", national_deaths.is_monotonic_increasing)
```

```python
# Threshold crossing dates
case_thresholds = [100, 500, 1000, 2000, 3500]
death_thresholds = [50, 100, 500, 1000, 1500]

print("Case threshold crossing dates:")
case_crossing_dates = {}
for threshold in case_thresholds:
    crossing = national_cases[national_cases >= threshold]
    case_crossing_dates[threshold] = crossing.index[0]
    print(f"  {threshold}: {crossing.index[0].date()}")

print("\nDeath threshold crossing dates:")
death_crossing_dates = {}
for threshold in death_thresholds:
    crossing = national_deaths[national_deaths >= threshold]
    death_crossing_dates[threshold] = crossing.index[0]
    print(f"  {threshold}: {crossing.index[0].date()}")
```
